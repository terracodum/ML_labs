## 3. Предобработка данных

Из-за того что большая часть моделей умеет воспринимать только числовые фичи, нам надо перевести категориальные данные в числовые.
- у нас имеется всего 2 столбца month, day
1. month это месяца --> логично перевести в числовое представление месяца
2. day это день недели --> переводим в дни недели с 1 по 7 начиная с понедельника 

```python
month_map = {'jan': 1, 'feb': 2, 'mar': 3, 'apr': 4, 'may': 5, 'jun': 6, 'jul': 7, 'aug': 8, 'sep': 9, 'oct': 10, 'nov': 11, 'dec': 12}
day_map = {'mon': 1, 'tue': 2, 'wed': 3, 'thu': 4, 'fri': 5,  'sat': 6, 'sun':7}
data['month'] = data['month'].map(month_map)
data['day'] = data['day'].map(day_map)
```

```python
data.head(20)
```

#### Можем еще раз проверить heatmap на возможность зависимости новых характеристик 

```python
corr_matrix = data.corr(numeric_only=True)

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix, 
    annot=True,     # Пишем значения внутри ячеек
    cmap='coolwarm', # Цветовая гамма (синий-белый-красный)
    fmt=".2f",      # Формат чисел (2 знака после запятой)
    vmin=-1, vmax=1 # Фиксируем границы цветов (от -1 до 1)
)
plt.show()
```

- отсюда видим что месяц коррелирует только с индексами влажности почвы
- дни вообще не влияют

## 4. Описательная статистика
- Анализ показал, что большинство признаков имеют асимметричные распределения, что согласуется с результатами визуального анализа. 
- Для признака area медианное значение близко к нулю, при наличии редких, но очень больших значений площади пожара. Это указывает на неравномерный характер распространения пожаров.

- Показатели пожарной опасности (FFMC, DMC, DC, ISI) характеризуются широким диапазоном значений и высокой вариативностью. 
- Наибольший разброс наблюдается у признака DC, что может свидетельствовать о влиянии длительных засушливых условий. 
- Метеорологические параметры (temp, RH, wind) имеют более стабильные статистические характеристики и меньший разброс значений, их графики более всех похожи на нормальное распределение. 
- В целом описательная статистика подтверждает наличие выбросов и неравномерность распределения площади лесных пожаров.
